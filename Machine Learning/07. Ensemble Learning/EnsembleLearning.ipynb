## Ensemble Learning

Ensemble learning is a machine learning approach where multiple models (called base learners) are combined to produce more accurate and robust predictions than any single model alone. This technique leverages the diversity of models to reduce variance, bias, and improve generalization on unseen data .

Key Types of Ensemble Methods

- Bagging (Bootstrap Aggregating): Trains base models **independently** on random subsets of data (with replacement) and aggregates predictions via majority voting (classification) or averaging (regression). Reduces variance and overfitting.

- Boosting: Trains models **sequentially**, with each new model focusing on correcting errors of the previous ones. Reduces bias and improves accuracy.

- Stacking: Trains diverse base models and uses their predictions as input features for a meta-model that learns the optimal combination strategy .

Mostly base learners are decision trees

EnsembleLearning uses --->multiple models(learners)---> combine all model predictions -->single prediction.


### Base Learners

 ## Bagging

- reduces variance

- here weak learners train on subset of data parallely
- In regression we take average of prediction
- In classification we take majority of class

```python
# Bagging with decision trees

from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize the base classifier (in this case, a decision tree)
base_classifier = DecisionTreeClassifier()

# Initialize the BaggingClassifier
bagging_classifier = BaggingClassifier(estimator=base_classifier, n_estimators=10, random_state=42)

# Train the BaggingClassifier
bagging_classifier.fit(X_train, y_train)

# Make predictions on the test set
y_pred = bagging_classifier.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## Boosting

- reduce bias
- In boosting weak learners train on data sequentially focusing on previous errors

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_digits
# Load dataset
X, y = load_digits(return_X_y=True)
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.25, random_state=23)
# Instantiate and train the model
gbc = GradientBoostingClassifier(n_estimators=300, learning_rate=0.05, random_state=100, max_features=5)
gbc.fit(train_X, train_y)
# Predict and evaluate
pred_y = gbc.predict(test_X)
acc = accuracy_score(test_y, pred_y)
print(f"Gradient Boosting Classifier accuracy is: {acc:.2f}")
```
