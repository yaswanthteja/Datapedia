## Ada Boost (adaptive Boosting)

AdaBoost, short for Adaptive Boosting, is an ensemble learning technique primarily used for classification tasks. It combines multiple weak learners to create a strong classifier, enhancing the accuracy of machine learning models. The algorithm was introduced by Freund and Schapire in 1997 and has since become a widely adopted technique for addressing binary classification challenges.

### How AdaBoost Works

The principle behind AdaBoost is to iteratively train weak classifiers on the training dataset, with each successive classifier giving more weight to the data points that are misclassified. The final AdaBoost model is a combination of all the weak classifiers, weighted according to their accuracies.

### Steps Involved in AdaBoost

1. Initialize Weights: For a dataset with N training data points, initialize N weights for each data point with equal values: `weight(xi) = 1/N`

2. Train Weak Classifiers: Train a weak classifier on the training dataset. The weak classifier should have an accuracy greater than 0.5, meaning it should perform better than a naive guess.

3. Calculate Error Rate and Importance: Calculate the error rate for each weak classifier and determine its importance using the formula: `alpha_k = 0.5 * ln((1 - error_k) / error_k)`

4. Update Weights: Update the weights assigned to each data point based on the accuracy of the weak classifier. The formula for updating the weights is: `w_i = w_i * exp(-alpha_k * y_i * M_k(x_i))`

5. Normalize Weights: Normalize the instance weights so that they sum up to 1: `W_i = W_i / sum(W)`

6. Repeat Steps 2-5: Repeat the above steps for a specified number of iterations or until a low training error is achieved.

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn import datasets, metrics


# Load dataset
iris = datasets.load_iris()

X = iris.data
y = iris.target

# Split dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

# Create AdaBoost classifier object
AdaBoostClf = AdaBoostClassifier(n_estimators=50, learning_rate=1)

# Train AdaBoost Classifier
model = AdaBoostClf.fit(X_train, y_train)

# Predict the response for test dataset
y_pred = model.predict(X_test)

# Model Accuracy
print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
```
